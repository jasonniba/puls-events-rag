Librairies

Charger les cellules

```python
import pandas as pd
import numpy as np
from langchain_huggingface import HuggingFaceEmbeddings

df = pd.read_csv("data/processed/openagenda_clean.csv")

print("Dimensions :", df.shape)
df[["Titre", "texte_rag"]].head()
```

Charger le modèle d'embeddings en local

```python
model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"batch_size": 32}
)
```

Tester la vectorisation sur 10 évènements

```python
textes_test = df["texte_rag"].fillna("").head(10).tolist()

vecteurs_test = model.embed_documents(textes_test)

print("Nombre de vecteurs :", len(vecteurs_test))
print("Dimension :", len(vecteurs_test[0]))
```

Générer les embeddings total

```python
textes = df["texte_rag"].fillna("").tolist()

vecteurs = model.embed_documents(textes)

embeddings = np.array(vecteurs, dtype="float32")

print("Dimensions des embeddings :", embeddings.shape)
```

```python
np.save("data/processed/embeddings_local.npy", embeddings)

print("Embeddings sauvegardés !")
```
